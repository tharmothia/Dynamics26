```python
import numpy as np
import matplotlib.pyplot as plt
# from scipy.integrate import solve_ivp
```

```python
alpha = 0.5   # parameter, 0 < alpha < 1
n = 1000      # number of terms to keep

x = np.linspace(-np.pi, np.pi, n)
y = np.zeros_like(x)

for k in range(n):
  y += 2**(-k*alpha) * np.cos(2**k * x)
```

```python
plt.plot(x,y)
plt.title(f"Weierstrass Function Approximation for n={n} Terms")
plt.xlabel("$-\pi < x < \pi$")
plt.ylabel(r"$W(x)=\sum_{k=0}^{n}2^{-n \alpha}\cos(2^nx)$")
plt.show()
```
