```python
from scipy.integrate import solve_ivp
import numpy as np
import matplotlib.pyplot as plt
import random
random.seed(42) # for reproducibility
```

```python
def centrifugal_governor(t, state, alpha, beta, epsilon):
  x, y, z = state

  x_dot = y
  y_dot = np.sin(x) * (z**2 * np.cos(x) - 1) - epsilon * y
  z_dot = alpha * (np.cos(x) - beta)

  return [x_dot, y_dot, z_dot]

# initial condition, small perturbation from eq
ic = [np.arccos(beta) + 0.01, 0, 1/np.sqrt(beta)]

alpha = 0.5
beta = 0.5
```

```python
epsilons_list = [2, 1, 0.5]

fig = plt.figure(figsize=(12, 10))
ax = fig.add_subplot(111, projection='3d')

# solve and plot system for various values of epsilon
for eps in epsilons_list:
    sol = solve_ivp(centrifugal_governor, (0, 500), ic, args=(alpha, beta, eps),
                    t_eval=np.linspace(0, 500, 5000))
    x, y, z = sol.y
    ax.plot3D(x, y, z, label=f'$\epsilon = {eps:.2f}$')

ax.plot3D(ic[0], ic[1], ic[2],
          'o', label='Initial Condition')

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title(rf'Centrifugal Governor for Various $\epsilon$, $t_f=500$')
ax.set_box_aspect(None, zoom=0.95) # so we can see the z-axis label
ax.legend()
plt.show()
```
