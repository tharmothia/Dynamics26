```python
import numpy as np
import matplotlib.pyplot as plt
# from scipy.integrate import solve_ivp
```

```python
alpha = 0.5   # parameter, 0 < alpha < 1
n = 1000      # number of terms to keep

x = np.linspace(-np.pi, np.pi, n)
y = np.zeros_like(x)

for k in range(n):
  y += 2**(-k*alpha) * np.cos(2**k * x)
```

```python
plt.plot(x,y)
plt.title(rf"Weierstrass Function Approximation for n={n} Terms")
plt.xlabel("$-\pi < x < \pi$")
plt.ylabel(r"$W(x)=\sum_{k=0}^{n}2^{-n \alpha}\cos(2^nx)$")
plt.show()
```

```python
# alpha = 0.5   # parameter, 0 < alpha < 1
n = 500      # number of terms to keep

x = np.linspace(-np.pi, np.pi, n)
y = np.zeros_like(x)

alpha = [0.1, 0.25, 0.5, 0.75, 0.9]

for a in alpha:
  for k in range(n):
    y += 2**(-k*a) * np.cos(2**k * x)

  plt.plot(x,y)
  plt.title(rf"Weierstrass Function Approximation for n={n} Terms with $\alpha$={a}")
  plt.xlabel("$-\pi < x < \pi$")
  plt.ylabel(r"$W(x)=\sum_{k=0}^{n}2^{-n \alpha}\cos(2^nx)$")
  plt.show()
```
